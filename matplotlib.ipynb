```python
#matplotlib simple line plot example
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 10)
y = x**2

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 45, 20, 15]

plt.pie(sizes, labels=labels)
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 45, 20, 15]
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 30, 40]

fig, ax = plt.subplots(1, 2)

ax[0].plot(x, y)
ax[0].set_title("Line")

ax[1].bar(x, y)
ax[1].set_title("Bar")

plt.show()
```
