```python
#matplotlib simple line plot example
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 10)
y = x**2

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.style.use('classic')
plt.show()
```

```python
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 30, 40]

fig, ax = plt.subplots(1, 2)

ax[0].plot(x, y)
ax[0].set_title("Line")

ax[1].bar(x, y)
ax[1].set_title("Bar")

plt.show()
```

```python
#object oriented approach to plotting
import matplotlib.pyplot as plt

# The assistant guesses what canvas to use
plt.plot([1, 2, 3], [10, 20, 15])
plt.title("My Plot")
plt.xlabel("X-Axis")
plt.show()
plt.legend()
```

```python
#advanced plotting with object oriented approach
import matplotlib.pyplot as plt

# 1. We explicitly create the Figure (canvas) and the Axes (the plot)
fig, ax = plt.subplots()

# 2. We tell THAT SPECIFIC plot to draw the line
ax.plot([1, 2, 3], [10, 20, 15])

# 3. We tell THAT SPECIFIC plot to set its title and labels
ax.set_title("My Plot")
ax.set_xlabel("X-Axis")
ax.set_ylabel("Y-Axis")
plt.style.use('ggplot')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)

# ---- First Plot ----
fig = plt.figure()

plt.plot(x, np.sin(x), '-')
plt.plot(x, np.cos(x), '--')

plt.title("Sine and Cosine Waves")
plt.legend(['Sine', 'Cosine'])

plt.show()

# ---- Subplots ----
fig2 = plt.figure()

plt.subplot(2, 1, 1)
plt.plot(x, np.sin(x))
plt.title("Sine")

plt.subplot(2, 1, 2)
plt.plot(x, np.cos(x))
plt.title("Cosine")

plt.tight_layout()
plt.show()

# ---- Save subplot figure ----
fig2.savefig('my_plot.png')
```
