```python
# Create a NumPy array of numbers 1–20
#Reshape it into (4, 5)
#Find:
       #Mean of each row
       #Max of each column
import numpy as np
a= np.arange(1,21);
print(a)
b=a.reshape(4,5)
print(b)
c = b.mean(axis=1) #Row axis=1
print("Mean of each row:", c)

d = b.max(axis=0) # Column axis=0
print("Max of each column:", d)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

total_sum = arr.sum()
n = arr.size

result = (total_sum - arr) / (n - 1)

print(result)
```

```python
#Normalize the array
import numpy as np
a= np.array([10,20,30,40,50])
norm=(a-np.min(a))/(np.max(a)-np.min(a))
print(norm)
```

```python
#Z-score Standardization
import numpy as np
a=np.array([10,20,30,40,50])
b=np.mean(a)
c=np.std(a)
z_score= (a-b)/c
print(z_score)
```
