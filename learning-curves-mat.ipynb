```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Generate some mock data (Simulating 10 epochs of training)
epochs = np.arange(1, 11)
train_loss = [0.9, 0.7, 0.5, 0.4, 0.3, 0.25, 0.2, 0.18, 0.15, 0.12]
val_loss =   [0.95, 0.8, 0.6, 0.55, 0.5, 0.48, 0.52, 0.55, 0.58, 0.6] # Overfitting at the end!

train_acc =  [0.6, 0.7, 0.8, 0.85, 0.9, 0.92, 0.94, 0.95, 0.96, 0.98]
val_acc =    [0.55, 0.65, 0.75, 0.8, 0.82, 0.83, 0.82, 0.81, 0.80, 0.79]

# 2. Initialize the Figure and Axes using the Object-Oriented API
# Creates a 1x2 grid (1 row, 2 columns). figsize is (width, height) in inches.
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# ---------------------------------------------------------
# Subplot 1: Loss Curve (Left)
# ---------------------------------------------------------
ax1.plot(epochs, train_loss, label='Training Loss', color='blue', marker='o', linestyle='-')
ax1.plot(epochs, val_loss, label='Validation Loss', color='red', marker='s', linestyle='--')

# Add context
ax1.set_title('Model Loss Over Time', fontsize=14, fontweight='bold')
ax1.set_xlabel('Epochs', fontsize=12)
ax1.set_ylabel('Loss (e.g., Cross-Entropy)', fontsize=12)
ax1.legend(loc='upper right') # Crucial for distinguishing lines
ax1.grid(True, alpha=0.3)      # Adds a subtle grid for readability

# ---------------------------------------------------------
# Subplot 2: Accuracy Curve (Right)
# ---------------------------------------------------------
ax2.plot(epochs, train_acc, label='Training Accuracy', color='green', marker='o', linestyle='-')
ax2.plot(epochs, val_acc, label='Validation Accuracy', color='orange', marker='s', linestyle='--')

# Add context
ax2.set_title('Model Accuracy Over Time', fontsize=14, fontweight='bold')
ax2.set_xlabel('Epochs', fontsize=12)
ax2.set_ylabel('Accuracy (%)', fontsize=12)
ax2.legend(loc='lower right')
ax2.grid(True, alpha=0.3)

# 3. Clean up the layout and display
plt.tight_layout() # Automatically adjusts padding so labels don't overlap
plt.show()

# Optional: Save the figure for a report
# fig.savefig('learning_curve.png', dpi=300, bbox_inches='tight')
```
